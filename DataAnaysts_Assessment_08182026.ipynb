```python
# Step 1: Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
# Step 2: Load dataset (upload your Excel file first in Colab)
#uploaded = files.upload()
#df = pd.read_excel("Online_Store_Orders.xlsx")

df = pd.read_excel("/content/Online_Store_Orders.xlsx")
df.columns = df.columns.str.strip()   # clean column names

# -------------------------------
# A. Identify Data Types
# -------------------------------
qual_cols = df.select_dtypes(include=['object']).columns.tolist()
quan_cols = df.select_dtypes(include=['number']).columns.tolist()

print("Qualitative Columns:", qual_cols)
print("Quantitative Columns:", quan_cols)

# -------------------------------
# B. Frequency Analysis
# -------------------------------
# Product Category
prod_freq = df['Product_Category'].value_counts(normalize=True).reset_index()
prod_freq.columns = ['Product_Category', 'Percentage']
prod_freq['Frequency'] = df['Product_Category'].value_counts().values
print("\nProduct Category Frequency:\n", prod_freq)

# City
city_freq = df['City'].value_counts(normalize=True).reset_index()
city_freq.columns = ['City', 'Percentage']
city_freq['Frequency'] = df['City'].value_counts().values
print("\nCity Frequency:\n", city_freq)

# Rating
rating_freq = df['Rating'].value_counts(normalize=True).reset_index()
rating_freq.columns = ['Rating', 'Percentage']
rating_freq['Frequency'] = df['Rating'].value_counts().values
print("\nRating Frequency:\n", rating_freq)

# Charts
prod_freq.plot(kind='bar', x='Product_Category', y='Frequency', title="Product Category Frequency")
plt.show()

city_freq.plot(kind='bar', x='City', y='Frequency', title="City Frequency")
plt.show()

# -------------------------------
# C. Central Tendency
# -------------------------------
print("\nOrder Value - Mean:", round(df['Order_Value'].mean(),2))
print("Order Value - Median:", round(df['Order_Value'].median(),2))
print("Order Value - Mode:", df['Order_Value'].mode()[0])

print("\nAge - Mean:", round(df['Age'].mean(),2))
print("Age - Median:", round(df['Age'].median(),2))
print("Age - Mode:", df['Age'].mode()[0])

print("\nDelivery Days - Mean:", round(df['Delivery_Days'].mean(),2))
print("Delivery Days - Median:", round(df['Delivery_Days'].median(),2))

# -------------------------------
# D. Percentile Analysis
# -------------------------------
print("\nOrder Value Percentiles:")
print("P25:", np.percentile(df['Order_Value'], 25))
print("P50 (Median):", np.percentile(df['Order_Value'], 50))
print("P75:", np.percentile(df['Order_Value'], 75))
print("P90:", np.percentile(df['Order_Value'], 90))

# -------------------------------
# E. IQR & Outlier Detection
# -------------------------------
Q1 = np.percentile(df['Order_Value'], 25)
Q3 = np.percentile(df['Order_Value'], 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
outliers = df[(df['Order_Value'] < lower_bound) | (df['Order_Value'] > upper_bound)]
print("\nOrder Value Outliers:\n", outliers[['Order_ID','Order_Value']])

# Delivery Days Outlier Detection
Q1_days = np.percentile(df['Delivery_Days'], 25)
Q3_days = np.percentile(df['Delivery_Days'], 75)
IQR_days = Q3_days - Q1_days
lower_days = Q1_days - 1.5*IQR_days
upper_days = Q3_days + 1.5*IQR_days
outliers_days = df[(df['Delivery_Days'] < lower_days) | (df['Delivery_Days'] > upper_days)]
print("\nDelivery Days Outliers:\n", outliers_days[['Order_ID','Delivery_Days']])

# -------------------------------
# F. Business Insights
# -------------------------------
print("\nBusiness Insights:")
print("1. Most popular product category:", df['Product_Category'].mode()[0])
print("2. City with strongest order activity:", df['City'].mode()[0])
print("3. Typical order value (median):", df['Order_Value'].median())
print("4. Outliers detected in Order Value:", len(outliers))
print("5. Recommendation: Focus marketing on", df['City'].mode()[0], "and stock more", df['Product_Category'].mode()[0])

# -------------------------------
# G. Final Management Summary
# -------------------------------

# Collect key findings
most_popular_category = df['Product_Category'].mode()[0]
top_city = df['City'].mode()[0]
median_order_value = df['Order_Value'].median()
p75_order_value = np.percentile(df['Order_Value'], 75)
num_outliers = len(outliers)

# Generate summary text
summary = (
    f"As a Junior Data Analyst, I observed that {top_city} has the strongest order activity "
    f"and {most_popular_category} is the most popular product category among customers. "
    f"The typical customer order value is around {median_order_value:.2f}, which represents the median "
    f"and is more reliable than the mean when extreme values are present. "
    f"The 75th percentile of order value is {p75_order_value:.2f}, meaning that 25% of orders are above this amount, "
    f"indicating a segment of high-value customers. "
    f"I identified {num_outliers} potential outlier orders, which may represent unusually large purchases "
    f"that could skew averages. "
    f"Based on these insights, I recommend focusing marketing efforts in {top_city} and ensuring sufficient stock "
    f"of {most_popular_category} products to meet demand."
)

print("\n--- Final Management Summary ---\n")
print(summary)
```

```python
import pandas as pd
import numpy as np
from datetime import date
from google.colab import files
from openpyxl import load_workbook
from openpyxl.styles import Alignment

# Load dataset
df = pd.read_excel("/content/Online_Store_Orders.xlsx")
df.columns = df.columns.str.strip()

# Create Excel writer
output_file = "Anitha_08182026_Data_Analysis_Assessment.xlsx"
writer = pd.ExcelWriter(output_file, engine='openpyxl')

# -------------------------------
# Raw Data
# -------------------------------
df.to_excel(writer, sheet_name="Raw Data", index=False)

# -------------------------------
# Frequency Analysis
# -------------------------------
prod_freq = df['Product_Category'].value_counts().reset_index()
prod_freq.columns = ['Product_Category', 'Frequency']
prod_freq['Percentage'] = (prod_freq['Frequency'] / len(df) * 100).round(2)

city_freq = df['City'].value_counts().reset_index()
city_freq.columns = ['City', 'Frequency']
city_freq['Percentage'] = (city_freq['Frequency'] / len(df) * 100).round(2)

rating_freq = df['Rating'].value_counts().reset_index()
rating_freq.columns = ['Rating', 'Frequency']
rating_freq['Percentage'] = (rating_freq['Frequency'] / len(df) * 100).round(2)

freq_df = pd.concat([
    prod_freq.assign(Table="Product Category"),
    city_freq.assign(Table="City"),
    rating_freq.assign(Table="Rating")
])
freq_df.to_excel(writer, sheet_name="Frequency Analysis", index=False)

# -------------------------------
# Central Tendency
# -------------------------------
central_tendency = pd.DataFrame({
    "Variable": ["Order_Value", "Age", "Delivery_Days"],
    "Mean": [df['Order_Value'].mean(), df['Age'].mean(), df['Delivery_Days'].mean()],
    "Median": [df['Order_Value'].median(), df['Age'].median(), df['Delivery_Days'].median()],
    "Mode": [df['Order_Value'].mode()[0], df['Age'].mode()[0], None]
})
central_tendency.to_excel(writer, sheet_name="Central Tendency", index=False)

# -------------------------------
# Percentile & IQR
# -------------------------------
percentiles = pd.DataFrame({
    "Percentile": ["P25", "P50", "P75", "P90"],
    "Order_Value": [
        np.percentile(df['Order_Value'], 25),
        np.percentile(df['Order_Value'], 50),
        np.percentile(df['Order_Value'], 75),
        np.percentile(df['Order_Value'], 90)
    ]
})

Q1 = np.percentile(df['Order_Value'], 25)
Q3 = np.percentile(df['Order_Value'], 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
outliers = df[(df['Order_Value'] < lower_bound) | (df['Order_Value'] > upper_bound)]

percentiles.to_excel(writer, sheet_name="Percentile & IQR", index=False, startrow=1)
outliers[['Order_ID','Order_Value']].to_excel(writer, sheet_name="Percentile & IQR", index=False, startrow=7)

# -------------------------------
# Insights
# -------------------------------
insights = [
    f"Assessment Title: Practical Data Analysis Assessment",
    f"Name: Anitha",
    f"Date: {date.today()}",
    f"Most popular product category: {df['Product_Category'].mode()[0]}",
    f"City with strongest order activity: {df['City'].mode()[0]}",
    f"Typical order value (median): {df['Order_Value'].median():.2f}",
    f"75th percentile of order value: {np.percentile(df['Order_Value'], 75):.2f}",
    f"Outliers detected in Order Value: {len(outliers)}"
]
pd.DataFrame(insights, columns=["Insights"]).to_excel(writer, sheet_name="Insights", index=False)

# -------------------------------
# Data Type Identification
# -------------------------------
data_types = pd.DataFrame({
    "Column": ["Order_ID","Customer","City","Age","Product_Category","Order_Value","Quantity","Delivery_Days","Rating"],
    "Qualitative/Quantitative": ["Qualitative","Qualitative","Qualitative","Quantitative","Qualitative","Quantitative","Quantitative","Quantitative","Quantitative"],
    "Discrete/Continuous": ["-","-","-","Continuous","-","Continuous","Discrete","Discrete","Discrete"],
    "Nominal/Ordinal": ["Nominal","Nominal","Nominal","-","Nominal","-","-","-","Ordinal"]
})
data_types.to_excel(writer, sheet_name="Data Types", index=False)

# -------------------------------
# Final Management Summary
# -------------------------------
summary = (
    f"As a Junior Data Analyst, I observed that {df['City'].mode()[0]} has the strongest order activity "
    f"and {df['Product_Category'].mode()[0]} is the most popular product category among customers. "
    f"The typical customer order value is around {df['Order_Value'].median():.2f}, which represents the median "
    f"and is more reliable than the mean when extreme values are present. "
    f"The 75th percentile of order value is {np.percentile(df['Order_Value'], 75):.2f}, meaning that 25% of orders are above this amount, "
    f"indicating a segment of high-value customers. "
    f"I identified {len(outliers)} potential outlier orders, which may represent unusually large purchases "
    f"that could skew averages. "
    f"Based on these insights, I recommend focusing marketing efforts in {df['City'].mode()[0]} and ensuring sufficient stock "
    f"of {df['Product_Category'].mode()[0]} products to meet demand."
)
pd.DataFrame([summary], columns=["Final Management Summary"]).to_excel(writer, sheet_name="Final Summary", index=False)

# -------------------------------
# Combined Calculations & Interpretations
# -------------------------------
combined = [
    ["Section","Calculation","Interpretation"],
    ["Frequency Analysis","Most popular category = "+df['Product_Category'].mode()[0],"This category has highest demand among customers."],
    ["Frequency Analysis","Top city = "+df['City'].mode()[0],"This city shows strongest order activity."],
    ["Central Tendency","Median Order Value = "+str(df['Order_Value'].median()),"Median is more reliable than mean when outliers exist."],
    ["Percentile","P75 = "+str(np.percentile(df['Order_Value'], 75)),"25% of orders are above this value, indicating high-value customers."],
    ["IQR","Outliers detected = "+str(len(outliers)),"These represent unusually large purchases that may skew averages."]
]
combined_df = pd.DataFrame(combined[1:], columns=combined[0])
combined_df.to_excel(writer, sheet_name="Calculations & Interpretations", index=False)

# Save workbook
writer.close()

# Wrap text in Final Summary sheet
wb = load_workbook(output_file)
ws = wb["Final Summary"]
for row in ws.iter_rows(min_row=2, max_row=2, min_col=1, max_col=1):
    for cell in row:
        cell.alignment = Alignment(wrap_text=True)
wb.save(output_file)

# Print summary in Colab
print("\n--- Final Management Summary ---\n")
print(summary)

# Download file
files.download(output_file)
```
