```python
import pandas as pd
from google.colab import files
#Upload the file and read the file
#uploaded = files.upload()
#df = pd.read_excel("Placement.xlsx")
#****************************************
#Read the file from the colab or drive directly
dataset = pd.read_excel("/content/Placement.xlsx")
#****************************************
print("Output:")
print("Total rows in the dataset:", dataset.shape[0])
print("Total Columns in the dataset:", dataset.shape[1])
column = 'ssc_p'
freqtable = pd.DataFrame(columns=["Unique_Values", "Frequency", "Relative_frequency", "Cumsum"])
print(freqtable)

freqtable["Unique_Values"] = dataset[column].value_counts().index
freqtable["Frequency"] = dataset[column].value_counts().values
freqtable["Relative_frequency"] = freqtable["Frequency"] / dataset.shape[0]
freqtable["Cumsum"] = freqtable["Relative_frequency"].cumsum()
#****************************************
#Sorting the data based on the Unique_values
freqtable = freqtable.sort_values(by="Unique_Values", ascending=True)
print(freqtable)
#****************************************
#Download the output in Excel format
output_file = "frequency_table_08182026.xlsx"
freqtable.to_excel(output_file, index=False)
files.download(output_file)


```
