```python
# Install required library
!pip install python-docx

from datetime import date
from docx import Document
from docx.shared import Pt

# Create Word document
doc = Document()
doc.add_heading("Practical Data Analysis Assessment", 0)
doc.add_paragraph("Name: Anitha")
doc.add_paragraph(f"Date: {date.today()}")

# -------------------------------
# Task dictionary: code + interpretation
# -------------------------------
tasks = {
    "Raw Data": {
        "code": "df.to_excel(writer, sheet_name='Raw Data', index=False)",
        "interpretation": "The raw dataset contains customer orders with details such as city, product category, order value, and rating. This forms the foundation for all further analysis."
    },
    "Frequency Analysis": {
        "code": """prod_freq = df['Product_Category'].value_counts().reset_index()
city_freq = df['City'].value_counts().reset_index()
rating_freq = df['Rating'].value_counts().reset_index()""",
        "interpretation": "Electronics is the most popular product category, while Chennai shows the strongest order activity. Ratings are mostly clustered around 4 and 5, indicating generally positive customer feedback."
    },
    "Central Tendency": {
        "code": """central_tendency = pd.DataFrame({
    "Variable": ["Order_Value", "Age", "Delivery_Days"],
    "Mean": [df['Order_Value'].mean(), df['Age'].mean(), df['Delivery_Days'].mean()],
    "Median": [df['Order_Value'].median(), df['Age'].median(), df['Delivery_Days'].median()],
    "Mode": [df['Order_Value'].mode()[0], df['Age'].mode()[0], None]
})""",
        "interpretation": "The median order value is more reliable than the mean when extreme values are present, showing typical customer spending behavior."
    },
    "Percentile & IQR": {
        "code": """Q1 = np.percentile(df['Order_Value'], 25)
Q3 = np.percentile(df['Order_Value'], 75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q3 + 1.5*IQR
outliers = df[(df['Order_Value'] < lower_bound) | (df['Order_Value'] > upper_bound)]""",
        "interpretation": "The 75th percentile highlights high-value customers, while IQR analysis detects unusually large purchases that may skew averages."
    },
    "Insights": {
        "code": """insights = [
    f"Most popular product category: {df['Product_Category'].mode()[0]}",
    f"City with strongest order activity: {df['City'].mode()[0]}",
    f"Typical order value (median): {df['Order_Value'].median():.2f}"
]""",
        "interpretation": "Summarizes key findings such as top category, top city, and typical order value, providing quick actionable insights."
    },
    "Data Types": {
        "code": """data_types = pd.DataFrame({
    "Column": ["Order_ID","Customer","City","Age","Product_Category","Order_Value","Quantity","Delivery_Days","Rating"],
    "Qualitative/Quantitative": ["Qualitative","Qualitative","Qualitative","Quantitative","Qualitative","Quantitative","Quantitative","Quantitative","Quantitative"],
    "Discrete/Continuous": ["Discrete","-","-","Continuous","-","Continuous","Discrete","Discrete","Discrete"],
    "Nominal/Ordinal": ["Nominal","Nominal","Nominal","-","Nominal","-","-","-","Ordinal"]
})""",
        "interpretation": "Classifies columns as qualitative or quantitative, discrete or continuous, and nominal or ordinal to guide statistical analysis."
    },
    "Final Management Summary": {
        "code": """summary = (
    f"As a Junior Data Analyst, I observed that {df['City'].mode()[0]} has the strongest order activity "
    f"and {df['Product_Category'].mode()[0]} is the most popular product category among customers."
)""",
        "interpretation": "Provides a concise business recommendation: focus marketing in Chennai and ensure sufficient stock of Electronics products."
    },
    "Charts": {
        "code": """plt.bar(prod_freq['Product_Category'], prod_freq['Frequency'])
plt.title("Product Category Frequency")
plt.show()""",
        "interpretation": "Visualizes frequency distributions, making trends easier to interpret and communicate to stakeholders."
    }
}

# -------------------------------
# Loop through tasks
# -------------------------------
for task_name, content in tasks.items():
    doc.add_heading(task_name, level=1)
    doc.add_paragraph("Code:")
    code_para = doc.add_paragraph(content["code"])
    code_para.style.font.name = 'Courier New'  # monospaced font
    code_para.style.font.size = Pt(10)
    doc.add_paragraph("Interpretation: " + content["interpretation"])

# Save Word document
output_doc = "Anitha_Data_Analysis_Assessment.docx"
doc.save(output_doc)

# Download Word file
from google.colab import files
files.download(output_doc)
```
